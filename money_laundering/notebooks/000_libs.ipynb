## Versões Instaladas

Verificando as versões das principais bibliotecas.

```python
# Testar importações
import sys
print("="*80)
print("VERIFICAÇÃO DE BIBLIOTECAS INSTALADAS")
print("="*80)

libs_to_check = {
    'pandas': 'pd',
    'numpy': 'np',
    'sklearn': 'sklearn',
    'scipy': 'scipy',
    'matplotlib': 'matplotlib',
    'seaborn': 'sns',
    'plotly': 'plotly',
    'pyspark': 'pyspark',
    'pyarrow': 'pyarrow',
    'xgboost': 'xgb',
    'lightgbm': 'lgb',
    'imblearn': 'imblearn',
    'category_encoders': 'category_encoders',
    'sqlalchemy': 'sqlalchemy',
    'dotenv': 'python-dotenv',
    'loguru': 'loguru',
    'tqdm': 'tqdm',
    'kaggle': 'kaggle',
    'requests': 'requests'
}

installed = []
not_installed = []

for lib, display_name in libs_to_check.items():
    try:
        __import__(lib)
        installed.append(display_name)
        print(f"✅ {display_name}")
    except ImportError:
        not_installed.append(display_name)
        print(f"❌ {display_name}")

print("\n" + "="*80)
print(f"RESUMO: {len(installed)}/{len(libs_to_check)} bibliotecas instaladas com sucesso")
print("="*80)

if not_installed:
    print(f"\n⚠️ Bibliotecas não instaladas: {', '.join(not_installed)}")
    print("Execute novamente as células de instalação acima.")
else:
    print("\n✅ Todas as bibliotecas foram instaladas com sucesso!")
    print("✅ Você está pronto para começar a trabalhar no projeto!")
```
